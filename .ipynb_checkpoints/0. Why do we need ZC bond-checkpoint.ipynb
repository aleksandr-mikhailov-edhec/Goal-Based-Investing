```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from tqdm import tqdm

import market_stochastic_simulation as mss
import retirement_client as pensioner
```

Imagine that an individual can pay 50.000 as a lump-sum payment. He needs to have 100.000 in 20 years. 

**Hypothesis:** Zero-Coupon bond can help him reach his objective with higher probability than if he had no ZC bond

To test this Hypothesis I set up stochastic simulations:
- vasichek process for st rates
- heston stochastic volatility for asset prices

I will compare results of 3 portfolios:
1) Portfolio of purely stocks
2) Portfolio that partly invests in ZC bond and is never rebalanced
3) Portfolio that partly invests in ladder bond index and is never rebalanced

## Set up Simulations

```python
# Parameters
S0 = 1000.0              # asset price
T = 20                  # time in years
N = 252*T              # number of time steps in simulation
M = 1000                # number of simulated paths
seed = 124              # simulation seed

simulated_market = mss.MarketSimulator(T,N,M,seed, S0, lambda_r = 0)
simulated_market.plot_market_simulation()
```
